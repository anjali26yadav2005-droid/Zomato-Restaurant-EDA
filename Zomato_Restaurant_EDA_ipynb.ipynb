```python
import matplotlib.pyplot as plt

# Count top 10 cities
city_counts = df['City'].value_counts().head(10)

# Create bar chart
plt.figure(figsize=(10, 5))
city_counts.plot(kind='bar')

# Add labels and title
plt.title('Top 10 Cities by Number of Restaurants')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')

# Show chart
plt.show()
```

```python
# Create line chart
plt.figure(figsize=(10, 5))
city_counts.plot(kind='line', marker='o')

# Add labels and title
plt.title('Number of Restaurants in Top 10 Cities')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')

# Show chart
plt.show()
```

```python
# Create histogram of restaurant ratings
plt.figure(figsize=(10, 5))
df['Aggregate rating'].plot(kind='hist', bins=10)

# Add labels and title
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Number of Restaurants')

# Show chart
plt.show()
```

```python
# Create scatter plot
plt.figure(figsize=(10, 5))
plt.scatter(df['Aggregate rating'], df['Votes'], alpha=0.5)

# Add labels and title
plt.title('Restaurant Ratings vs Number of Votes')
plt.xlabel('Aggregate Rating')
plt.ylabel('Number of Votes')

# Show chart
plt.show()
```

```python
# Count online delivery options
delivery_counts = df['Has Online delivery'].value_counts()

# Create pie chart
plt.figure(figsize=(7, 7))
plt.pie(
    delivery_counts,
    labels=delivery_counts.index,
    autopct='%1.1f%%'
)

# Add title
plt.title('Restaurants with Online Delivery')

# Show chart
plt.show()
```

```python
import seaborn as sns

# Select numerical columns
numeric_data = df.select_dtypes(include='number')

# Create heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(numeric_data.corr(), annot=True)

# Add title
plt.title('Correlation Heatmap of Zomato Data')

# Show chart
plt.show()
```
